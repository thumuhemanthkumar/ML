```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Sample data
X = np.array([
    [1, 3], [1, 2], [2, 3],
    [8, 8], [9, 8], [8, 9],
    [4, 6], [5, 6], [6, 5],
    [10, 10], [11, 10], [10, 11],
    [12, 12], [13, 12], [12, 13],
    [14, 14], [15, 14], [14, 15],
    [16, 16], [17, 16], [16, 17]
])

# Create K-Means model
kmeans = KMeans(n_clusters=6, random_state=42, n_init=10)

# Fit the model
kmeans.fit(X)

# Get cluster labels
labels = kmeans.labels_

# Get cluster centers
centers = kmeans.cluster_centers_

print("Cluster Labels:")
print(labels)

print("\nCluster Centers:")
for center in centers:
    print(f"{center[0]:.2f},{center[1]:.2f}")

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels)

# Plot centroids
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker='X',
    s=100
)

plt.title("K-Means Clustering")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
